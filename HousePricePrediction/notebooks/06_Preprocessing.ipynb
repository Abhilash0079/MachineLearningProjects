# <b>House Price Prediction</b>

## <b>Phase 6 – Data Preprocessing</b>


### <b>Objective</b>

The goal of this notebook is to prepare the dataset for machine learning.

Tasks covered:

- Load feature-engineered dataset
- Split features and target
- Train-test split
- Feature scaling
- Save preprocessing pipeline
- Save processed datasets

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import joblib
```

```python
df = pd.read_csv("../data/processed/Housing_Feature_Engineered.csv")
```

```python
print(df.shape)

df.head()
```

```python
X = df.drop("price", axis=1)

y = df["price"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Target   :", y_train.shape)
print("Testing Target    :", y_test.shape)
```

```python
numerical_columns = X_train.select_dtypes(
    include=["int64", "float64"]
).columns

print(numerical_columns)
```

```python
preprocessing_pipeline = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)
```

```python
X_train_scaled = preprocessing_pipeline.fit_transform(X_train)
```

```python
X_test_scaled = preprocessing_pipeline.transform(X_test)
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)
```

```python
X_train_scaled.head()
```

```python
joblib.dump(
    preprocessing_pipeline,
    "../models/preprocessing_pipeline.pkl"
)
```

```python
X_train_scaled.to_csv(
    "../data/processed/X_train.csv",
    index=False
)

X_test_scaled.to_csv(
    "../data/processed/X_test.csv",
    index=False
)

y_train.to_csv(
    "../data/processed/y_train.csv",
    index=False
)

y_test.to_csv(
    "../data/processed/y_test.csv",
    index=False
)
```

#### <b>Data Preprocessing Summary</b>

##### Train-Test Split

- Training Data: 80%
- Testing Data: 20%

##### Feature Scaling

- StandardScaler applied.
- Scaler fitted only on training data.
- Test data transformed using the fitted scaler.

##### Pipeline

A preprocessing pipeline was created and saved for reuse during model training and deployment.
