# <b>House Price Prediction</b>

## <b>Phase 5 – Feature Engineering</b>

### <b>Objective</b>

The goal of this notebook is to prepare the cleaned dataset for machine learning by converting categorical variables into numerical representations.

Tasks covered:

- Load cleaned dataset
- Separate features and target
- Identify categorical and numerical features
- Encode categorical variables
- Verify transformed dataset
- Save engineered dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/processed/Housing_Cleaned.csv')
```

```python
df_fe = df.copy()
```

```python
print("Dataset Shape:", df_fe.shape)

df_fe.head()
```

```python
X = df_fe.drop("price", axis=1)

y = df_fe["price"]
```

```python
numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_columns = X.select_dtypes(
    include="str"
).columns

print("Numerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)
```

```python
binary_columns = [
    "mainroad",
    "guestroom",
    "basement",
    "hotwaterheating",
    "airconditioning",
    "prefarea"
]

for column in binary_columns:
    df_fe[column] = df_fe[column].map({
        "yes": 1,
        "no": 0
    })
```

```python
df_fe = pd.get_dummies(
    df_fe,
    columns=["furnishingstatus"],
    drop_first=True,
    dtype=int
)
```

```python
print(df_fe.isnull().sum())
```

```python
df_fe.to_csv(
    "../data/processed/Housing_Feature_Engineered.csv",
    index=False
)
```

#### <b>Feature Engineering Summary</b>

##### Binary Features

The following binary categorical variables were encoded:

- mainroad
- guestroom
- basement
- hotwaterheating
- airconditioning
- prefarea

Encoding:

- yes → 1
- no → 0

##### Multi-Class Feature

The `furnishingstatus` column was transformed using one-hot encoding.

##### <b>Outcome</b>

The dataset now contains only numerical features, making it suitable for machine learning algorithms.
