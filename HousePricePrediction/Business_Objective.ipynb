# <b>Phase 0 — Business Understanding</b>

## <b>Objective</b>

Before writing any code, we first understand why we are building the model. This mirrors how real data science projects begin in industry.

## <b>Business Problem</b>

A real estate company wants to estimate the selling price of a house before listing it for sale. Currently, prices are determined manually, which can lead to inconsistent and inaccurate estimates.

The company wants a machine learning solution that predicts the expected selling price based on house characteristics such as area, number of bedrooms, bathrooms, parking availability, furnishing status, location-related features, and other property attributes.

## <b>Business Objective</b>

<b>Build a regression model that predicts house prices accurately so the company can:</b>

- Price properties more consistently.
- Reduce manual effort and human bias.
- Help buyers and sellers make informed decisions.
- Support data-driven pricing strategies.
- Machine Learning Problem

<b>Learning Type: Supervised Learning</b>

- Problem Type: Regression
- Target Variable: price
- Inputs (Features)

<b>Typical features include:</b>

- Area
- Bedrooms
- Bathrooms
- Stories
- Main Road
- Guest Room
- Basement
- Hot Water Heating
- Air Conditioning
- Parking
- Preferred Area
- Furnishing Status
- Price.

<b>Expected Output</b>

A trained model that predicts a continuous house price for a new property based on its features.

## <b>Success Criteria</b>

The project will be considered successful if:

- The model generalizes well to unseen data.
- Regression metrics (MAE, RMSE, and R²) indicate good predictive performance.
- The model is explainable and suitable for deployment.
- The solution can be integrated into a simple web application.

## <b>Business Constraints</b>
- Predictions should be generated quickly.
- The model should be interpretable enough for business users.
- Data quality directly affects prediction accuracy.
- The solution should work for new houses with similar feature distributions as the training data.
