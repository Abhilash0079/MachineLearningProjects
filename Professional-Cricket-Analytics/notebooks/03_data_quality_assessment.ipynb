#### <b>❓ Missing Value Analysis</b>

Missing values are common in real-world datasets.

However, not every missing value represents poor data quality.

For example:

- A delivery without a wicket will naturally have no dismissed player.
- Some matches may not have a recorded city.
- Player of the Match may not exist for certain abandoned matches.

Understanding *why* values are missing is more important than simply counting them.

```python
matches_df.isnull().sum().sort_values(ascending=False)
```

```python
deliveries_df.isnull().sum().sort_values(ascending=False)
```

```python
players_df.isnull().sum().sort_values(ascending=False)
```

##### <b>💡 Business Insight</b>

Missing values should always be investigated before applying any imputation or deletion strategy.

Domain knowledge plays a critical role in determining whether missing values indicate:

- Expected business scenarios
- Data collection issues
- Processing errors

Blindly removing rows with missing values can result in the loss of valuable information.

#### <b>📊 Missing Value Percentage</b>

Absolute counts can be misleading.

For example:

- 10 missing values out of 10 million records are insignificant.
- 10 missing values out of 20 records are highly concerning.

Therefore, percentages provide better context.

```python
def missing_percentage(df):

    missing = (
        df.isnull().sum() / len(df)
    ) * 100

    return (
        missing[missing > 0]
        .sort_values(ascending=False)
        .round(2)
    )
```

```python
missing_percentage(matches_df)
```

```python
missing_percentage(deliveries_df)
```

```python
missing_percentage(players_df)
```

#### <b>✅ Business Rule Validation</b>

One important validation rule in cricket analytics is:

**Total Runs = Batter Runs + Extra Runs**

If this relationship is violated, the dataset may contain processing errors.

Business rule validation helps ensure data integrity before advanced analysis.

```python
invalid_rows = deliveries_df[
    deliveries_df["total_runs"] !=
    (
        deliveries_df["batter_runs"] +
        deliveries_df["extra_runs"]
    )
]

print("Invalid rows:", len(invalid_rows))
```
