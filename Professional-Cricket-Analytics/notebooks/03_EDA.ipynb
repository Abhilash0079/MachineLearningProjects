Blindly removing rows with missing values can result in the loss of valuable information.

#### <b>📊 Missing Value Percentage</b>

Absolute counts can be misleading.

For example:

- 10 missing values out of 10 million records are insignificant.
- 10 missing values out of 20 records are highly concerning.

Therefore, percentages provide better context.

```python
def missing_percentage(df):

    missing = (
        df.isnull().sum() / len(df)
    ) * 100

    return (
        missing[missing > 0]
        .sort_values(ascending=False)
        .round(2)
    )
```

```python
missing_percentage(matches_df)
```

```python
missing_percentage(deliveries_df)
```

```python
missing_percentage(players_df)
```

#### <b>✅ Business Rule Validation</b>

One important validation rule in cricket analytics is:

**Total Runs = Batter Runs + Extra Runs**

If this relationship is violated, the dataset may contain processing errors.

Business rule validation helps ensure data integrity before advanced analysis.

```python
invalid_rows = deliveries_df[
    deliveries_df["total_runs"] !=
    (
        deliveries_df["batter_runs"] +
        deliveries_df["extra_runs"]
    )
]

print("Invalid rows:", len(invalid_rows))
```

#### <b>🏏 Matches Played by Each Team</b>

##### Business Question

Which IPL teams have played the highest number of matches?

Understanding participation helps compare teams fairly before evaluating their success.

```python
team1 = matches_df[["team_1"]].rename(
    columns={"team_1": "Team"}
)

team2 = matches_df[["team_2"]].rename(
    columns={"team_2": "Team"}
)

all_teams = pd.concat(
    [team1, team2],
    ignore_index=True
)

matches_played = (
    all_teams["Team"]
    .value_counts()
    .reset_index()
)

matches_played.columns = [
    "Team",
    "Matches Played"
]

matches_played
```

```python
fig = px.bar(
    matches_played,
    x="Team",
    y="Matches Played",
    color="Matches Played",
    text="Matches Played",
    title="Total Matches Played by Each IPL Team"
)

fig.update_layout(
    xaxis_title="Team",
    yaxis_title="Matches Played"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Teams that joined the IPL in later seasons naturally have fewer matches.

Therefore, comparing only total wins may be misleading. Win percentage provides a fairer comparison.

#### <b>🏆 Total Wins by Team</b>

##### Business Question

Which IPL teams have won the highest number of matches?

```python
team_wins = (
    matches_df["winner"]
    .dropna()
    .value_counts()
    .reset_index()
)

team_wins.columns = [
    "Team",
    "Wins"
]

team_wins
```

```python
fig = px.bar(
    team_wins,
    x="Wins",
    y="Team",
    orientation="h",
    color="Wins",
    text="Wins",
    title="Total IPL Wins by Team"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Teams with consistently high win counts have generally maintained stronger squads, effective leadership, and stable management over multiple seasons.

#### <b>📈 Team Winning Percentage</b>

##### Business Question

Which team has been the most successful based on win percentage?

Win percentage is a more balanced metric because it considers the number of matches played.

```python
team_summary = (
    matches_played
    .merge(team_wins, on="Team", how="left")
)

team_summary["Wins"] = (
    team_summary["Wins"]
    .fillna(0)
)

team_summary["Win Percentage"] = (
    team_summary["Wins"]
    /
    team_summary["Matches Played"]
    * 100
).round(2)

team_summary.sort_values(
    "Win Percentage",
    ascending=False,
    inplace=True
)

team_summary
```

```python
fig = px.bar(
    team_summary,
    x="Team",
    y="Win Percentage",
    color="Win Percentage",
    text="Win Percentage",
    title="IPL Team Winning Percentage"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Win percentage provides a normalized measure of team performance and allows fair comparison between teams that have played different numbers of matches.

#### <b>📅 Team Wins by Season</b>

##### Business Question

How has each team's performance changed across IPL seasons?

This analysis helps identify periods of dominance, decline, or consistent performance.

```python
season_team_wins = (
    matches_df
    .dropna(subset=["winner"])
    .groupby(
        ["season", "winner"]
    )
    .size()
    .reset_index(name="Wins")
)

season_team_wins.head()
```

```python
fig = px.line(
    season_team_wins,
    x="season",
    y="Wins",
    color="winner",
    markers=True,
    title="Season-wise Wins by Team"
)

fig.update_layout(
    xaxis_title="Season",
    yaxis_title="Wins"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Season-wise performance highlights long-term trends.

Teams may experience:

- Consistent dominance
- Temporary decline
- Strong comeback seasons
- Performance fluctuations after squad changes

#### <b>🪙 Toss Winner vs Match Winner</b>

##### Business Question

Does winning the toss increase the probability of winning the match?

This analysis evaluates whether toss advantage translates into match victories.

```python
matches_df["Toss Won Match"] = (
    matches_df["toss_winner"]
    ==
    matches_df["winner"]
)

toss_impact = (
    matches_df["Toss Won Match"]
    .value_counts()
    .reset_index()
)

toss_impact.columns = [
    "Won Toss & Match",
    "Matches"
]

toss_impact
```

```python
fig = px.pie(
    toss_impact,
    names="Won Toss & Match",
    values="Matches",
    hole=0.45,
    title="Impact of Winning the Toss"
)

fig.show()
```

##### <b>💡 Business Insight</b>

If a large proportion of toss winners also win the match, it suggests that external factors such as pitch conditions or dew may significantly influence match outcomes.

However, if the proportions are similar, team performance may have a greater impact than the toss itself.
