#### <b>📊 Missing Value Percentage</b>

Absolute counts can be misleading.

For example:

- 10 missing values out of 10 million records are insignificant.
- 10 missing values out of 20 records are highly concerning.

Therefore, percentages provide better context.

```python
def missing_percentage(df):

    missing = (
        df.isnull().sum() / len(df)
    ) * 100

    return (
        missing[missing > 0]
        .sort_values(ascending=False)
        .round(2)
    )
```

```python
missing_percentage(deliveries_df)
```

```python
missing_percentage(players_df)
```

#### <b>📈 Team Winning Percentage</b>

##### Business Question

Which team has been the most successful based on win percentage?

Win percentage is a more balanced metric because it considers the number of matches played.

```python
team_summary = (
    matches_played
    .merge(team_wins, on="Team", how="left")
)

team_summary["Wins"] = (
    team_summary["Wins"]
    .fillna(0)
)

team_summary["Win Percentage"] = (
    team_summary["Wins"]
    /
    team_summary["Matches Played"]
    * 100
).round(2)

team_summary.sort_values(
    "Win Percentage",
    ascending=False,
    inplace=True
)

team_summary
```

#### <b>🏏 Most Runs Scored</b>

Which batters have scored the highest number of runs in IPL history?

Total runs provide a basic measure of batting productivity and longevity.

```python
batter_runs = (
    deliveries_df
    .groupby("batter", as_index=False)["batter_runs"]
    .sum()
    .sort_values("batter_runs", ascending=False)
)

batter_runs.rename(
    columns={"batter_runs": "Runs"},
    inplace=True
)

batter_runs.head(10)
```

```python
fig = px.bar(
    batter_runs.head(10),
    x="Runs",
    y="batter",
    orientation="h",
    text="Runs",
    color="Runs",
    title="Top 10 Run Scorers in IPL"
)

fig.update_layout(
    xaxis_title="Runs",
    yaxis_title="Batter"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Batters with high run totals generally demonstrate:

- Longevity
- Consistency
- Regular team selection
- Ability to perform across multiple seasons

However, total runs alone do not indicate batting efficiency. Strike rate and average should also be considered.

#### <b>🔴 Balls Faced</b>

##### Business Question

How many legal deliveries has each batter faced?

Balls faced is a fundamental metric used to calculate strike rate and understand batting workload.

```python
balls_faced = (
    deliveries_df[
        deliveries_df["is_legal_delivery"] == True
    ]
    .groupby("batter", as_index=False)
    .size()
)

balls_faced.rename(
    columns={"size": "Balls Faced"},
    inplace=True
)

balls_faced.head()
```

#### <b>⚡ Strike Rate</b>

##### Business Question

Which batters score runs most quickly?

Strike Rate measures the number of runs scored per 100 legal deliveries.

```python
strike_rate = batter_runs.merge(
    balls_faced,
    on="batter"
)

strike_rate["Strike Rate"] = (
    strike_rate["Runs"]
    /
    strike_rate["Balls Faced"]
    * 100
).round(2)

strike_rate.sort_values(
    "Strike Rate",
    ascending=False,
    inplace=True
)

strike_rate.head(10)
```

> **Important:** Batters with very few balls faced can have extremely high strike rates that are not representative of sustained performance. In professional analysis, a minimum balls-faced threshold (e.g., 200 or 500 balls) is often applied before comparing strike rates.

```python
qualified_sr = strike_rate[
    strike_rate["Balls Faced"] >= 500
]

qualified_sr.head(10)
```

```python
fig = px.bar(
    qualified_sr.head(10),
    x="Strike Rate",
    y="batter",
    orientation="h",
    color="Strike Rate",
    text="Strike Rate",
    title="Top Strike Rates (Minimum 500 Balls Faced)"
)

fig.show()
```

##### <b>💡 Business Insight</b>

A high strike rate indicates a batter's ability to score quickly, making such players particularly valuable during powerplays and death overs. Applying a minimum balls-faced criterion ensures that comparisons are statistically meaningful.

#### <b>0️⃣ Dot Ball Percentage</b>

##### Business Question

Which batters consume the highest percentage of dot balls?

Dot Ball Percentage measures how often a batter fails to score from a legal delivery.

A lower dot ball percentage generally indicates better strike rotation and pressure management.

```python
legal_deliveries = deliveries_df[
    deliveries_df["is_legal_delivery"] == True
]
```

```python
dot_balls = (
    legal_deliveries[
        legal_deliveries["batter_runs"] == 0
    ]
    .groupby("batter")
    .size()
    .reset_index(name="Dot Balls")
)

dot_balls.head()
```

```python
dot_ball_percentage = (
    balls_faced
    .merge(dot_balls, on="batter", how="left")
)

dot_ball_percentage["Dot Balls"] = (
    dot_ball_percentage["Dot Balls"]
    .fillna(0)
)
```

```python
dot_ball_percentage["Dot Ball %"] = (
    dot_ball_percentage["Dot Balls"]
    /
    dot_ball_percentage["Balls Faced"]
    *100
).round(2)

dot_ball_percentage.head()
```

```python
qualified_dot = dot_ball_percentage[
    dot_ball_percentage["Balls Faced"] >= 500
]

qualified_dot = qualified_dot.sort_values(
    "Dot Ball %",
    ascending=True
)

fig = px.bar(
    qualified_dot.head(10),
    x="Dot Ball %",
    y="batter",
    orientation="h",
    color="Dot Ball %",
    text="Dot Ball %",
    title="Lowest Dot Ball Percentage (Minimum 500 Balls)"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Batters with lower dot-ball percentages are generally better at rotating the strike.

These players:

- Keep the scoreboard moving
- Reduce pressure
- Build partnerships
- Force bowlers to change their plans

#### <b>💥 Boundary Percentage</b>

##### Business Question

Which batters score the highest proportion of their runs through boundaries?

Boundary Percentage indicates how aggressively a batter scores.

```python
boundaries = (
    legal_deliveries[
        legal_deliveries["is_boundary"] == True
    ]
    .groupby("batter")
    .size()
    .reset_index(name="Boundary Balls")
)

boundaries.head()
```

```python
boundary_percentage = (
    balls_faced
    .merge(boundaries, on="batter", how="left")
)

boundary_percentage["Boundary Balls"] = (
    boundary_percentage["Boundary Balls"]
    .fillna(0)
)
```

```python
boundary_percentage["Boundary %"] = (
    boundary_percentage["Boundary Balls"]
    /
    boundary_percentage["Balls Faced"]
    *100
).round(2)

boundary_percentage.head()
```

```python
qualified_boundary = boundary_percentage[
    boundary_percentage["Balls Faced"] >= 500
]

qualified_boundary = qualified_boundary.sort_values(
    "Boundary %",
    ascending=False
)

fig = px.bar(
    qualified_boundary.head(10),
    x="Boundary %",
    y="batter",
    orientation="h",
    color="Boundary %",
    text="Boundary %",
    title="Highest Boundary Percentage"
)

fig.show()
```

##### <b>💡 Business Insight</b>

A high boundary percentage suggests an aggressive batting style.

However, aggression should be interpreted alongside consistency and strike rotation to assess overall batting quality.

#### <b>🏃 Run Distribution</b>

##### Business Question

How does each batter accumulate runs?

Do they rely on:

- Singles?
- Doubles?
- Triples?
- Boundaries?

```python
run_distribution = (
    legal_deliveries
    .groupby(
        ["batter", "batter_runs"]
    )
    .size()
    .unstack(fill_value=0)
)

run_distribution.head()
```

```python
run_distribution.columns = [
    f"{col}_Run" for col in run_distribution.columns
]

run_distribution.reset_index(inplace=True)

run_distribution.head()
```

#### <b>📈 Batting Average</b>

##### Business Question

Which batters score the most runs before getting dismissed?

Batting Average is calculated as:

Average = Total Runs ÷ Number of Times Out

Not-out innings are excluded from the denominator.

```python
dismissals = (
    legal_deliveries[
        legal_deliveries["player_dismissed"] ==
        legal_deliveries["batter"]
    ]
    .groupby("batter")
    .size()
    .reset_index(name="Dismissals")
)

dismissals.head()
```

```python
batting_average = (
    batter_runs
    .merge(dismissals, on="batter", how="left")
)

batting_average["Dismissals"] = (
    batting_average["Dismissals"]
    .fillna(0)
)
```

```python
batting_average = batting_average[
    batting_average["Dismissals"] > 0
]
```

```python
batting_average["Average"] = (
    batting_average["Runs"]
    /
    batting_average["Dismissals"]
).round(2)

batting_average.sort_values(
    "Average",
    ascending=False,
    inplace=True
)

batting_average.head(10)
```

> **Important:** Similar to strike rate, batting averages can be misleading for players with very few dismissals. In professional analysis, analysts often apply a minimum qualification (e.g., 500 balls faced or 20 innings) before ranking players.

```python
qualified_average = batting_average.merge(
    balls_faced,
    on="batter"
)

qualified_average = qualified_average[
    qualified_average["Balls Faced"] >= 500
]

qualified_average = qualified_average.sort_values(
    "Average",
    ascending=False
)

fig = px.bar(
    qualified_average.head(10),
    x="Average",
    y="batter",
    orientation="h",
    color="Average",
    text="Average",
    title="Highest Batting Average (Minimum 500 Balls)"
)

fig.show()
```

##### 💡 Business Insight

Batting average reflects a player's ability to remain at the crease while scoring runs.

Players with consistently high averages often play anchor roles and contribute to building stable innings.

#### <b>💰 Runs Conceded</b>

##### Business Question

How many runs has each bowler conceded?

Runs conceded form the basis for calculating economy rate and bowling average.

```python
bowler_runs = (
    deliveries_df
    .groupby("bowler", as_index=False)["total_runs"]
    .sum()
)

bowler_runs.rename(
    columns={"total_runs": "Runs Conceded"},
    inplace=True
)

bowler_runs.head()
```

#### <b>⚽ Balls Bowled</b>

##### Business Question

How many legal deliveries has each bowler bowled?

Only legal deliveries are considered when calculating overs, economy rate, and bowling strike rate.

```python
balls_bowled = (
    deliveries_df[
        deliveries_df["is_legal_delivery"] == True
    ]
    .groupby("bowler", as_index=False)
    .size()
)

balls_bowled.rename(
    columns={"size": "Balls Bowled"},
    inplace=True
)

balls_bowled.head()
```

#### <b>🏏 Overs Bowled</b>

Overs bowled are calculated from legal deliveries.

Formula:

Overs = Balls Bowled ÷ 6

```python
balls_bowled["Overs"] = (
    balls_bowled["Balls Bowled"] / 6
).round(1)

balls_bowled.head()
```

#### <b>📉 Economy Rate</b>

##### Business Question

Which bowlers concede the fewest runs per over?

Economy Rate = Runs Conceded ÷ Overs Bowled

This metric is particularly important in T20 cricket, where restricting runs is often as valuable as taking wickets.

```python
economy = (
    bowler_runs
    .merge(
        balls_bowled[
            ["bowler", "Balls Bowled", "Overs"]
        ],
        on="bowler"
    )
)

economy["Economy"] = (
    economy["Runs Conceded"]
    /
    economy["Overs"]
).round(2)

economy.head()
```

> **Important:** Economy rate should only be compared among bowlers who have bowled a meaningful number of overs. Analysts often set a minimum qualification, such as 300 legal deliveries or 50 overs, to avoid misleading comparisons from small sample sizes.

```python
qualified_economy = economy[
    economy["Balls Bowled"] >= 300
]

qualified_economy = qualified_economy.sort_values(
    "Economy"
)

qualified_economy.head(10)
```

```python
fig = px.bar(
    qualified_economy.head(10),
    x="Economy",
    y="bowler",
    orientation="h",
    color="Economy",
    text="Economy",
    title="Best Economy Rate (Minimum 300 Balls Bowled)"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Economical bowlers are invaluable in T20 cricket because they consistently apply pressure by limiting scoring opportunities. Even without taking many wickets, they can influence match outcomes by forcing batters into risky shots.

#### <b>📊 Bowling Average</b>

##### Business Question

Which bowlers concede the fewest runs for every wicket they take?

Bowling Average measures the number of runs conceded per wicket.

Formula:

Bowling Average = Runs Conceded ÷ Wickets

A lower bowling average generally indicates a more effective wicket-taking bowler.

```python
bowling_average = (
    economy.merge(
        bowler_wickets,
        on="bowler",
        how="left"
    )
)

bowling_average["Wickets"] = (
    bowling_average["Wickets"]
    .fillna(0)
)

bowling_average = bowling_average[
    bowling_average["Wickets"] > 0
]

bowling_average["Bowling Average"] = (
    bowling_average["Runs Conceded"]
    /
    bowling_average["Wickets"]
).round(2)

bowling_average.head()
```

```python
qualified_average = bowling_average[
    bowling_average["Balls Bowled"] >= 300
].sort_values("Bowling Average")
```

```python
fig = px.bar(
    qualified_average.head(10),
    x="Bowling Average",
    y="bowler",
    orientation="h",
    color="Bowling Average",
    text="Bowling Average",
    title="Best Bowling Average (Minimum 300 Balls)"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Bowling average reflects how efficiently a bowler converts runs conceded into wickets.

Lower averages indicate bowlers who regularly dismiss batters without leaking excessive runs.

#### <b>⚡ Bowling Strike Rate</b>

##### Business Question

How frequently does a bowler take wickets?

Formula:

Strike Rate = Balls Bowled ÷ Wickets

Lower values indicate bowlers who take wickets more frequently.

```python
bowling_sr = (
    bowling_average.copy()
)

bowling_sr["Bowling Strike Rate"] = (
    bowling_sr["Balls Bowled"]
    /
    bowling_sr["Wickets"]
).round(2)

bowling_sr.head()
```

```python
qualified_sr = bowling_sr[
    bowling_sr["Balls Bowled"] >= 300
].sort_values("Bowling Strike Rate")

fig = px.bar(
    qualified_sr.head(10),
    x="Bowling Strike Rate",
    y="bowler",
    orientation="h",
    color="Bowling Strike Rate",
    text="Bowling Strike Rate",
    title="Best Bowling Strike Rate"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Strike Rate measures wicket-taking frequency.

Bowlers with lower strike rates regularly break partnerships and prevent batters from settling.

#### <b>🎯 Dot Ball Percentage</b>

##### Business Question

Which bowlers create the most pressure by bowling dot balls?

Dot balls force batters into taking risks, increasing wicket opportunities.

```python
dot_balls = (
    deliveries_df[
        (deliveries_df["is_legal_delivery"] == True)
        &
        (deliveries_df["total_runs"] == 0)
    ]
    .groupby("bowler")
    .size()
    .reset_index(name="Dot Balls")
)

dot_balls.head()
```

#### <b>⏱️ Bowling Performance Across Match Phases</b>

##### Business Question

How do bowlers perform during different phases of the innings?

Phases:

- Powerplay (Overs 1–6)
- Middle Overs (Overs 7–15)
- Death Overs (Overs 16–20)

Different phases demand different bowling skills and strategies.

```python
phase_df = legal_deliveries.copy()
```

```python
phase_df["Phase"] = np.select(
    [
        phase_df["over"] <= 6,
        phase_df["over"].between(7, 15),
        phase_df["over"] >= 16
    ],
    [
        "Powerplay",
        "Middle Overs",
        "Death Overs"
    ],
    default="Other"
)
```

```python
phase_summary = (
    phase_df
    .groupby(["Phase", "bowler"])
    .agg(
        Balls=("ball", "count"),
        Runs=("total_runs", "sum"),
        Wickets=("is_bowler_wicket", "sum")
    )
    .reset_index()
)

phase_summary.head()
```

```python
phase_summary["Overs"] = (
    phase_summary["Balls"] / 6
).round(2)

phase_summary["Economy"] = (
    phase_summary["Runs"]
    /
    phase_summary["Overs"]
).round(2)

phase_summary["Strike Rate"] = (
    phase_summary["Balls"]
    /
    phase_summary["Wickets"]
).replace(np.inf, np.nan).round(2)

phase_summary.head()
```

```python
fig = px.box(
    phase_summary,
    x="Phase",
    y="Economy",
    color="Phase",
    title="Economy Rate Across Match Phases"
)

fig.show()
```

##### <b>💡 Business Insight</b>

Powerplay specialists usually focus on early wickets.

Middle-over bowlers aim to control the scoring rate.

Death-over specialists balance wicket-taking with run prevention under high-pressure situations.

#### <b>🏃 Chasing Versus Defending Success</b>

##### Business Question

Do IPL teams win more frequently while chasing or while defending a target?

A match is classified as:

- `Chasing Win` when the second-innings batting team wins
- `Defending Win` when the first-innings batting team wins
- `No Result / Tie` when no standard winner is available

```python
first_team = (
    innings_summary[
        innings_summary["innings"] == 1
    ][
        [
            "match_id",
            "batting_team",
            "Innings_Runs"
        ]
    ]
    .rename(
        columns={
            "batting_team": "First_Innings_Team",
            "Innings_Runs": "First_Innings_Runs"
        }
    )
)

second_team = (
    innings_summary[
        innings_summary["innings"] == 2
    ][
        [
            "match_id",
            "batting_team",
            "Innings_Runs"
        ]
    ]
    .rename(
        columns={
            "batting_team": "Second_Innings_Team",
            "Innings_Runs": "Second_Innings_Runs"
        }
    )
)
```

```python
match_result_context = (
    matches_df[
        [
            "match_id",
            "season",
            "venue",
            "winner",
            "toss_winner",
            "toss_decision"
        ]
    ]
    .merge(
        first_team,
        on="match_id",
        how="left"
    )
    .merge(
        second_team,
        on="match_id",
        how="left"
    )
)

match_result_context.head()
```

```python
conditions = [
    (
        match_result_context["winner"]
        ==
        match_result_context["Second_Innings_Team"]
    ),
    (
        match_result_context["winner"]
        ==
        match_result_context["First_Innings_Team"]
    )
]

choices = [
    "Chasing Win",
    "Defending Win"
]

match_result_context["Winning_Method"] = np.select(
    conditions,
    choices,
    default="No Result / Tie"
)
```

```python
winning_method_summary = (
    match_result_context["Winning_Method"]
    .value_counts()
    .reset_index()
)

winning_method_summary.columns = [
    "Winning Method",
    "Matches"
]

winning_method_summary
```

```python
fig = px.pie(
    winning_method_summary,
    names="Winning Method",
    values="Matches",
    hole=0.4,
    title="Chasing Versus Defending Success"
)

fig.show()
```

##### <b>💡 Business Insight</b>

The chasing-versus-defending split indicates whether batting second provides a general advantage.

Possible explanations include:

- Dew
- Improved knowledge of the target
- Changing pitch conditions
- Pressure while defending

```python
venue_results = (
    match_result_context[
        match_result_context["Winning_Method"].isin(
            [
                "Chasing Win",
                "Defending Win"
            ]
        )
    ]
    .groupby(
        [
            "venue",
            "Winning_Method"
        ]
    )
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

venue_results.head()
```

```python
if "Chasing Win" not in venue_results.columns:
    venue_results["Chasing Win"] = 0

if "Defending Win" not in venue_results.columns:
    venue_results["Defending Win"] = 0

venue_results["Completed Matches"] = (
    venue_results["Chasing Win"]
    +
    venue_results["Defending Win"]
)

venue_results["Chasing Win %"] = (
    venue_results["Chasing Win"]
    /
    venue_results["Completed Matches"]
    * 100
).round(2)
```

#### <b>🤝 Partnership Identification</b>

##### Business Question

How can we uniquely identify a batting partnership?

A batting partnership consists of two players batting together.

```python
partnership_df["Partnership"] = (
    partnership_df[
        ["batter", "non_striker"]
    ]
    .apply(
        lambda players: " & ".join(sorted(players)),
        axis=1
    )
)

partnership_df.head()
```

#### <b>⭐ Partnership Consistency</b>

##### Business Question

Which batting pairs regularly produce meaningful partnerships?

A pair should not be considered reliable because of only one exceptional innings.

For this analysis, a meaningful partnership is defined as one that scores at least 30 runs.

##### Consistency Index

Consistency Index = 30+ Partnerships ÷ Innings Together × 100

```python
partnership_consistency = (
    match_partnerships
    .groupby(
        "Partnership",
        as_index=False
    )
    .agg(
        Innings_Together=(
            "match_id",
            "count"
        ),
        Total_Runs=(
            "Partnership_Runs",
            "sum"
        ),
        Average_Partnership=(
            "Partnership_Runs",
            "mean"
        ),
        Thirty_Plus=(
            "Partnership_Runs",
            lambda scores: (
                scores >= 30
            ).sum()
        ),
        Fifty_Plus=(
            "Partnership_Runs",
            lambda scores: (
                scores >= 50
            ).sum()
        ),
        Hundred_Plus=(
            "Partnership_Runs",
            lambda scores: (
                scores >= 100
            ).sum()
        )
    )
)
```

```python
partnership_consistency["Consistency Index"] = (
    partnership_consistency["Thirty_Plus"]
    /
    partnership_consistency["Innings_Together"]
    * 100
).round(2)

partnership_consistency[
    "Average_Partnership"
] = partnership_consistency[
    "Average_Partnership"
].round(2)
```

```python
qualified_consistency = (
    partnership_consistency[
        partnership_consistency[
            "Innings_Together"
        ] >= 10
    ]
    .sort_values(
        "Consistency Index",
        ascending=False
    )
)

qualified_consistency.head(10)
```

```python
fig = px.bar(
    qualified_consistency.head(10),
    x="Consistency Index",
    y="Partnership",
    orientation="h",
    color="Consistency Index",
    text="Consistency Index",
    hover_data=[
        "Innings_Together",
        "Average_Partnership",
        "Thirty_Plus",
        "Fifty_Plus"
    ],
    title="Most Consistent IPL Batting Partnerships"
)

fig.update_layout(
    xaxis_title="Consistency Index (%)",
    yaxis_title="Partnership",
    yaxis={
        "categoryorder": "total ascending"
    }
)

fig.show()
```

##### <b>💡 Business Insight</b>

The Consistency Index separates regularly productive partnerships from pairs whose career totals are driven by only a few exceptional innings.

A reliable partnership provides stability and reduces dependency on individual performances.

#### <b>🕸️ Partnership Network Analysis</b>

##### Business Question

How are batters connected through successful partnerships?

In this network:

- Each node represents a batter
- Each edge represents a batting partnership
- Edge thickness represents total partnership runs
- Larger nodes represent greater partnership involvement

Network analysis helps identify central players who build productive combinations with multiple teammates.

```python
network_data = partnership_runs.copy()

network_data[
    [
        "Player_1",
        "Player_2"
    ]
] = network_data[
    "Partnership"
].str.split(
    " & ",
    n=1,
    expand=True
)
```

```python
network_data = network_data[
    network_data["Runs"] >= 500
].copy()
```

```python
partnership_graph = nx.Graph()
```

```python
for _, row in network_data.iterrows():
    partnership_graph.add_edge(
        row["Player_1"],
        row["Player_2"],
        weight=row["Runs"]
    )
```

```python
positions = nx.spring_layout(
    partnership_graph,
    seed=42,
    k=0.8
)
```

```python
edge_x = []
edge_y = []

for player_1, player_2 in (
    partnership_graph.edges()
):
    x0, y0 = positions[player_1]
    x1, y1 = positions[player_2]

    edge_x.extend(
        [
            x0,
            x1,
            None
        ]
    )

    edge_y.extend(
        [
            y0,
            y1,
            None
        ]
    )
```

#### <b>⭐ Death-Over Finisher Score</b>

##### Business Question

How can multiple death-over batting metrics be combined into one score?

The Finisher Score combines:

- Strike rate
- Boundary percentage
- Low dot-ball percentage
- Run volume

The score is designed for comparative analysis, not as an official cricket statistic.

```python
finisher_df = qualified_death_batters.copy()
```

```python
def min_max_scale(series):
    minimum = series.min()
    maximum = series.max()

    if maximum == minimum:
        return pd.Series(
            100,
            index=series.index
        )

    return (
        (series - minimum)
        /
        (maximum - minimum)
        * 100
    )
```

```python
finisher_df["SR_Score"] = min_max_scale(
    finisher_df["Strike Rate"]
)

finisher_df["Boundary_Score"] = min_max_scale(
    finisher_df["Boundary %"]
)

finisher_df["Run_Score"] = min_max_scale(
    finisher_df["Runs"]
)

finisher_df["Dot_Control_Score"] = (
    100
    -
    min_max_scale(
        finisher_df["Dot Ball %"]
    )
)
```

```python
finisher_df["Finisher Score"] = (
    finisher_df["SR_Score"] * 0.35
    +
    finisher_df["Boundary_Score"] * 0.25
    +
    finisher_df["Dot_Control_Score"] * 0.20
    +
    finisher_df["Run_Score"] * 0.20
).round(2)
```

```python
top_finishers = (
    finisher_df
    .sort_values(
        "Finisher Score",
        ascending=False
    )
    .head(10)
)

top_finishers[
    [
        "batter",
        "Runs",
        "Balls",
        "Strike Rate",
        "Boundary %",
        "Dot Ball %",
        "Finisher Score"
    ]
]
```

```python
fig = px.bar(
    top_finishers,
    x="Finisher Score",
    y="batter",
    orientation="h",
    color="Finisher Score",
    text="Finisher Score",
    hover_data=[
        "Runs",
        "Balls",
        "Strike Rate",
        "Boundary %",
        "Dot Ball %"
    ],
    title="Top IPL Finishers"
)

fig.update_layout(
    yaxis={
        "categoryorder": "total ascending"
    }
)

fig.show()
```

#### <b>📈 Required Run Rate Analysis</b>

##### Business Question

How does chase pressure change during the second innings?

Required Run Rate is calculated as:

Required Run Rate = Runs Required ÷ Overs Remaining

This metric shows how quickly the chasing team must score to reach the target.

```python
chase_df = pressure_df[
    pressure_df["innings"] == 2
].copy()
```

```python
chase_df["Runs_Before_Ball"] = (
    chase_df
    .groupby("match_id")["total_runs"]
    .cumsum()
    -
    chase_df["total_runs"]
)
```

```python
chase_df["Legal_Balls_Before"] = (
    chase_df
    .groupby("match_id")["is_legal_delivery"]
    .cumsum()
    -
    chase_df["is_legal_delivery"]
)
```

```python
chase_df["Runs_Required"] = (
    chase_df["Target"]
    -
    chase_df["Runs_Before_Ball"]
)

chase_df["Balls_Remaining"] = (
    120
    -
    chase_df["Legal_Balls_Before"]
)
```

```python
chase_df["Required Run Rate"] = np.where(
    chase_df["Balls_Remaining"] > 0,
    (
        chase_df["Runs_Required"]
        /
        chase_df["Balls_Remaining"]
        * 6
    ),
    np.nan
).round(2)
```

```python
chase_df["Current Run Rate"] = np.where(
    chase_df["Legal_Balls_Before"] > 0,
    (
        chase_df["Runs_Before_Ball"]
        /
        chase_df["Legal_Balls_Before"]
        * 6
    ),
    0
).round(2)
```

```python
chase_df["Run Rate Pressure"] = (
    chase_df["Required Run Rate"]
    -
    chase_df["Current Run Rate"]
).round(2)
```

> The calculation assumes a normal 20-over innings. Rain-reduced matches and revised targets require the actual innings-ball limit and official revised target.

```python
qualified_pressure_batters = (
    pressure_batter_stats[
        pressure_batter_stats["Balls"] >= 100
    ]
    .sort_values(
        "Strike Rate",
        ascending=False
    )
)

qualified_pressure_batters.head(10)
```

#### <b>📈 Match Momentum Analysis</b>

##### Business Question

How does scoring momentum change throughout an innings?

Momentum is approximated using over-by-over:

- Runs
- Wickets
- Cumulative score
- Run rate

This allows analysts to identify acceleration, collapses, and turning points.

```python
over_momentum = (
    phase_df
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team",
            "over"
        ],
        as_index=False
    )
    .agg(
        Over_Runs=("total_runs", "sum"),
        Wickets=("is_wicket", "sum"),
        Legal_Balls=("is_legal_delivery", "sum")
    )
)
```

```python
over_momentum = over_momentum.sort_values(
    [
        "match_id",
        "innings",
        "over"
    ]
)

over_momentum["Cumulative_Runs"] = (
    over_momentum
    .groupby(
        [
            "match_id",
            "innings"
        ]
    )["Over_Runs"]
    .cumsum()
)

over_momentum["Cumulative_Wickets"] = (
    over_momentum
    .groupby(
        [
            "match_id",
            "innings"
        ]
    )["Wickets"]
    .cumsum()
)
```

```python
over_momentum["Cumulative_Balls"] = (
    over_momentum
    .groupby(
        [
            "match_id",
            "innings"
        ]
    )["Legal_Balls"]
    .cumsum()
)

over_momentum["Cumulative Run Rate"] = np.where(
    over_momentum["Cumulative_Balls"] > 0,
    (
        over_momentum["Cumulative_Runs"]
        /
        over_momentum["Cumulative_Balls"]
        * 6
    ),
    np.nan
).round(2)
```

#### <b>⚠️ Limitations of Pressure Analytics</b>

This analysis uses practical proxies for pressure, but pressure cannot be measured perfectly from scorecards alone.

Important limitations include:

1. Required run rate does not capture wickets remaining by itself.

2. Match pressure differs based on batter quality, bowling quality, venue, pitch, and tournament stage.

3. Rain-reduced matches require revised targets and revised ball limits.

4. A high required run rate early in an innings may be less critical than the same rate in the final two overs.

5. Close-match thresholds are subjective.

6. Career totals favour players with longer careers.

7. Death-over performance may be influenced by the quality of the opposing team.

8. A batter may score quickly after the match outcome is already nearly certain.

Future modeling should create a ball-level win probability and measure how each delivery changes the probability of victory.

```python
def safe_divide(numerator, denominator, multiplier=1):
    """
    Safely divide two numeric values.

    Returns NaN when the denominator is zero.
    """

    if denominator == 0:
        return np.nan

    return (numerator / denominator) * multiplier
```

```python
batter_summary = (
    deliveries_df
    .groupby("batter", as_index=False)
    .agg(
        Runs=("batter_runs", "sum"),
        Balls=("is_legal_delivery", "sum"),
        Fours=("is_four", "sum"),
        Sixes=("is_six", "sum"),
        Boundaries=("is_boundary", "sum"),
        Dot_Balls=("is_dot_ball", "sum")
    )
)
```

```python
batter_dismissals = (
    deliveries_df[
        deliveries_df["player_dismissed"].notna()
    ]
    .groupby(
        "player_dismissed",
        as_index=False
    )
    .size()
    .rename(
        columns={
            "player_dismissed": "batter",
            "size": "Dismissals"
        }
    )
)

batter_summary = batter_summary.merge(
    batter_dismissals,
    on="batter",
    how="left"
)

batter_summary["Dismissals"] = (
    batter_summary["Dismissals"]
    .fillna(0)
)
```

```python
batter_summary["Strike Rate"] = np.where(
    batter_summary["Balls"] > 0,
    batter_summary["Runs"]
    /
    batter_summary["Balls"]
    * 100,
    np.nan
).round(2)

batter_summary["Batting Average"] = np.where(
    batter_summary["Dismissals"] > 0,
    batter_summary["Runs"]
    /
    batter_summary["Dismissals"],
    np.nan
).round(2)

batter_summary["Dot Ball %"] = np.where(
    batter_summary["Balls"] > 0,
    batter_summary["Dot_Balls"]
    /
    batter_summary["Balls"]
    * 100,
    np.nan
).round(2)

batter_summary["Boundary %"] = np.where(
    batter_summary["Balls"] > 0,
    batter_summary["Boundaries"]
    /
    batter_summary["Balls"]
    * 100,
    np.nan
).round(2)
```

```python
top_run_scorers = (
    batter_summary
    .sort_values(
        "Runs",
        ascending=False
    )
    .head(10)
)

top_run_scorers[
    [
        "batter",
        "Runs",
        "Balls",
        "Strike Rate",
        "Batting Average",
        "Fours",
        "Sixes",
        "Dot Ball %"
    ]
]
```

```python
qualified_batters = (
    batter_summary[
        batter_summary["Balls"] >= 500
    ]
    .copy()
)

qualified_batters["Impact Score"] = (
    (
        qualified_batters["Strike Rate"].rank(
            pct=True
        )
        * 35
    )
    +
    (
        qualified_batters["Batting Average"].rank(
            pct=True
        )
        * 30
    )
    +
    (
        qualified_batters["Runs"].rank(
            pct=True
        )
        * 25
    )
    +
    (
        (
            1
            -
            qualified_batters["Dot Ball %"].rank(
                pct=True
            )
        )
        * 10
    )
).round(2)

top_impact_batters = (
    qualified_batters
    .sort_values(
        "Impact Score",
        ascending=False
    )
    .head(10)
)

top_impact_batters[
    [
        "batter",
        "Runs",
        "Batting Average",
        "Strike Rate",
        "Dot Ball %",
        "Impact Score"
    ]
]
```

The best batter cannot be identified using only total runs.

A balanced batter assessment should distinguish among:

- **Wicket preservation:** ability to maintain a strong average
- **Pressure management:** ability to limit dot balls
- **Boundary production:** ability to accelerate the innings

This distinction is important during auctions because different batting roles require different combinations of these qualities.

```python
first_innings_by_venue = (
    deliveries_df[
        deliveries_df["innings"] == 1
    ]
    .groupby(
        [
            "match_id",
            "batting_team"
        ],
        as_index=False
    )
    .agg(
        First_Innings_Score=("total_runs", "sum")
    )
)

first_innings_by_venue = (
    first_innings_by_venue
    .merge(
        matches_df[
            [
                "match_id",
                "venue"
            ]
        ],
        on="match_id",
        how="left"
    )
)
```

```python
venue_score_summary = (
    first_innings_by_venue
    .groupby("venue", as_index=False)
    .agg(
        Matches=("match_id", "nunique"),
        Average_First_Innings_Score=(
            "First_Innings_Score",
            "mean"
        ),
        Highest_First_Innings_Score=(
            "First_Innings_Score",
            "max"
        ),
        Lowest_First_Innings_Score=(
            "First_Innings_Score",
            "min"
        )
    )
)

venue_score_summary[
    "Average_First_Innings_Score"
] = venue_score_summary[
    "Average_First_Innings_Score"
].round(2)
```

```python
venue_chase_summary = (
    chase_results
    .merge(
        matches_df[
            [
                "match_id",
                "venue"
            ]
        ],
        on="match_id",
        how="left"
    )
    .groupby("venue", as_index=False)
    .agg(
        Chases=("match_id", "nunique"),
        Successful_Chases=("Chase_Won", "sum")
    )
)

venue_chase_summary["Chasing Success %"] = np.where(
    venue_chase_summary["Chases"] > 0,
    venue_chase_summary["Successful_Chases"]
    /
    venue_chase_summary["Chases"]
    * 100,
    np.nan
).round(2)
```

The thresholds are analytical assumptions rather than official IPL classifications.

```python
executive_matches_df = matches_df.copy()

executive_matches_df["Is_Close_Match"] = (
    executive_matches_df["win_by_runs"]
    .between(1, 10)
    |
    executive_matches_df["win_by_wickets"]
    .between(1, 3)
)

close_matches_summary = executive_matches_df[
    executive_matches_df["Is_Close_Match"]
].copy()

print(
    "Total close matches:",
    close_matches_summary["match_id"].nunique()
)

print(
    "Percentage of matches classified as close:",
    round(
        close_matches_summary["match_id"].nunique()
        /
        executive_matches_df["match_id"].nunique()
        * 100,
        2
    ),
    "%"
)
```

```python
close_team_1 = close_matches_summary[
    [
        "match_id",
        "team_1",
        "winner"
    ]
].rename(
    columns={
        "team_1": "team"
    }
)

close_team_2 = close_matches_summary[
    [
        "match_id",
        "team_2",
        "winner"
    ]
].rename(
    columns={
        "team_2": "team"
    }
)

close_team_records = pd.concat(
    [
        close_team_1,
        close_team_2
    ],
    ignore_index=True
)

close_team_records["Win"] = (
    close_team_records["team"]
    ==
    close_team_records["winner"]
).astype(int)
```

```python
close_team_performance = (
    close_team_records
    .groupby("team", as_index=False)
    .agg(
        Close_Matches=("match_id", "nunique"),
        Close_Wins=("Win", "sum")
    )
)

close_team_performance["Close Match Win %"] = (
    close_team_performance["Close_Wins"]
    /
    close_team_performance["Close_Matches"]
    * 100
).round(2)

qualified_close_teams = (
    close_team_performance[
        close_team_performance["Close_Matches"] >= 10
    ]
    .sort_values(
        "Close Match Win %",
        ascending=False
    )
)

qualified_close_teams
```

#### <b>🏏 Strategic Batting Recommendations</b>

Based on the completed batter, partnership, venue and phase analyses, a franchise should structure its batting strategy around clearly defined roles.

##### <b>1. Opening Strategy</b>

Opening batters should be selected using:

- Powerplay strike rate
- Boundary percentage
- Dot-ball control
- Performance against the new ball
- Partnership compatibility

The highest career run scorer is not automatically the best Powerplay opener.

##### <b>2. Middle-Over Strategy</b>

Middle-order batters should demonstrate:

- Wicket preservation
- Strike rotation
- Ability to score against spin
- Partnership stability
- Capacity to accelerate before the Death Overs

##### <b>3. Finishing Strategy</b>

Finishers should be selected using death-over statistics rather than overall career strike rate.

Important measures include:

- Death-over strike rate
- Boundary percentage
- Dot-ball percentage
- Runs scored in the final five overs
- Performance in close matches
- Sample size

##### <b>4. Batting-Order Flexibility</b>

A fixed batting order may not be optimal for every match.

The order can be adjusted according to:

- Venue
- Required run rate
- Match phase
- Bowling matchups
- Wickets remaining
- Player form

##### <b>5. Partnership Planning</b>

Historical partnership data can identify combinations that produce both stability and scoring efficiency.

Teams should consider preserving successful combinations while also accounting for left-right batting balance and opposition matchups.

#### <b>🎯 Strategic Bowling Recommendations</b>

Bowling resources should be allocated according to phase-specific strengths.

##### <b>1. Powerplay Bowling</b>

Powerplay bowlers should be evaluated using:

- New-ball wickets
- Dot-ball percentage
- Economy
- Boundary concession
- Ability to dismiss opening batters

##### <b>2. Middle-Over Bowling</b>

Middle-over specialists should provide:

- Wicket-taking ability
- Run-rate control
- Pressure through dot balls
- Effective performance against established batters

##### <b>3. Death-Over Bowling</b>

Death bowlers should be selected using death-over data rather than overall economy.

Important indicators include:

- Death-over economy
- Death-over wickets
- Bowling strike rate
- Dot-ball percentage
- Boundary concession percentage
- Close-match performance

##### <b>4. Bowling-Unit Balance</b>

A strong bowling unit should contain complementary profiles:

- New-ball wicket-taker
- Powerplay control bowler
- Middle-over wicket-taker
- Middle-over containment bowler
- Death-over specialist
- Flexible sixth-bowling option

##### <b>5. Matchup-Based Usage</b>

Bowling changes should consider batter-specific records, venue conditions and the current match state rather than following a fixed over allocation in every game.

#### <b>🏟️ Venue and Toss Recommendations</b>

Historical venue analysis can support toss and match-strategy decisions.

##### <b>At High-Scoring, Chase-Friendly Venues</b>

A franchise may:

- Prefer chasing
- Retain additional finishing depth
- Avoid depending only on a par score
- Protect death-over bowling resources
- Plan for a higher target

##### <b>At High-Scoring, Defence-Friendly Venues</b>

A franchise may:

- Consider batting first
- Maximise Powerplay scoring
- Build a score above the historical average
- Select bowlers capable of defending totals
- Plan matchups for the final five overs

##### <b>At Lower-Scoring Venues</b>

A franchise may:

- Prioritise wicket preservation
- Select high-control bowlers
- Avoid excessive early risk
- Treat every partnership as strategically important
- Value fielding more heavily

##### <b>Important Limitation</b>

Historical venue data should support—not replace—current information such as:

- Pitch report
- Weather
- Dew
- Boundary dimensions
- Surface preparation
- Recent venue behaviour

#### <b>💼 Auction and Retention Recommendations</b>

The analysis can support evidence-based squad planning.

##### <b>1. Avoid Evaluating Players Only Through Career Totals</b>

Career totals are influenced by the number of matches played. Efficiency, role and recent performance must also be considered.

##### <b>2. Prioritise Scarce Tactical Roles</b>

Specialists who can perform difficult roles may be more valuable than another general top-order batter.

Scarce roles can include:

- Reliable death-over finishers
- Death-over bowlers
- Powerplay wicket-takers
- Middle-over spin hitters
- Flexible all-rounders
- High-quality wicketkeepers

##### <b>3. Use Qualification Thresholds</b>

An exceptional metric from a very small sample should not be treated as proven performance.

Minimum matches, innings, balls or overs should be used before ranking players.

##### <b>4. Combine Performance and Role Fit</b>

A player may be statistically strong but still duplicate an existing role in the squad.

Auction decisions should consider:

- Performance
- Tactical role
- Squad balance
- Venue suitability
- Overseas-player restrictions
- Availability
- Age and development potential
- Expected acquisition cost

##### <b>5. Retain Complementary Player Combinations</b>

Strong partnerships and complementary player profiles can be retained together when the combination produces stable performance across multiple seasons.

##### <b>6. Avoid Overpaying for Unsustainable Performance</b>

A single successful season may not represent long-term quality.

Franchises should compare:

- Career performance
- Recent form
- Season-to-season consistency
- Role-specific performance
- Opposition strength
- Venue-adjusted performance

#### <b>⚠️ Project Limitations</b>

Although the project provides extensive historical analysis, the following limitations should be recognised.

##### <b>1. Scorecard-Based Data</b>

The dataset primarily contains scorecard and ball-by-ball information. It does not include detailed tracking information.

##### <b>2. No Ball-Tracking Data</b>

The analysis cannot directly measure:

- Ball speed
- Ball trajectory
- Swing
- Seam movement
- Spin rate
- Release point
- Shot direction
- Bat speed

##### <b>3. Limited Fielding Measurement</b>

The dataset records dismissal involvement but not:

- Runs saved
- Dropped catches
- Missed chances
- Fielding difficulty
- Throw quality

##### <b>4. No Player Fitness Information</b>

Injuries, fatigue, workload and fitness status are not represented.

##### <b>5. No Salary or Auction-Price Data</b>

The analysis evaluates cricket performance but cannot directly calculate value for money.

##### <b>6. No Detailed Weather Data</b>

Temperature, humidity, rain probability and dew are not included.

##### <b>7. No Detailed Pitch Data</b>

The same venue may produce different surfaces across seasons and matches.

##### <b>8. Historical Team Names</b>

Franchise renaming and restructuring may create separate team categories unless names are standardised.

##### <b>9. Career Aggregation Bias</b>

Career totals favour players and teams with longer participation histories.

##### <b>10. Small-Sample Bias</b>

Some players may appear exceptional because they played only a small number of deliveries or matches.

##### <b>11. Subjective Composite Scores</b>

Finisher, consistency and impact-score weights are analytical assumptions rather than official cricket measures.

##### <b>12. Close-Match Definition</b>

The thresholds used to classify close matches are subjective.

##### <b>13. Rain-Reduced Matches</b>

Standard required-run-rate calculations may require special handling for revised targets and reduced innings.

##### <b>14. Correlation Is Not Causation</b>

Historical relationships do not prove that one factor directly caused the match result.
