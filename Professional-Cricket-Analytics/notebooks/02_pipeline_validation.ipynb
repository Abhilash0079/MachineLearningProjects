#### Import Libraries

```python
from pathlib import Path
import pandas as pd
```

#### Processed Paths

```python
processed_folder = Path("../data/processed")

matches_path = processed_folder / "matches.csv"
deliveries_path = processed_folder / "deliveries.csv"
players_path = processed_folder / "players.csv"
```

#### Load Datasets

```python
matches_df = pd.read_csv(matches_path,parse_dates=["match_date"])
deliveries_df = pd.read_csv(deliveries_path,parse_dates=["match_date"])
players_df = pd.read_csv(players_path)
```

#### Display Shape

```python
print("Matches shape:", matches_df.shape)
print("Deliveries shape:", deliveries_df.shape)
print("Players shape:", players_df.shape)
```

#### Sample Data

```python
print(f"Matches Data:\n{matches_df.head()}")
print(f"Deliveries Data:\n{deliveries_df.head()}")
print(f"Players Data:\n{players_df.head()}")
```

#### Missing Data

```python
matches_df.isnull().sum().sort_values(
    ascending=False
)
```

```python
deliveries_df.isnull().sum().sort_values(
    ascending=False
).head(15)
```

```python
players_df.isnull().sum()
```

#### Check Match IDs

```python
print(
    "Unique matches in matches dataset:",
    matches_df["match_id"].nunique()
)

print(
    "Unique matches in deliveries dataset:",
    deliveries_df["match_id"].nunique()
)

print(
    "Unique matches in players dataset:",
    players_df["match_id"].nunique()
)
```

#### Check Duplicates

```python
print(
    "Duplicate match IDs:",
    matches_df["match_id"].duplicated().sum()
)
```

```python
print(
    "Duplicate player participation records:",
    players_df.duplicated(
        subset=[
            "match_id",
            "team",
            "player_name",
        ]
    ).sum()
)
```

```python
deliveries_df.duplicated()
```

#### Validate Scores

```python
deliveries_df[
    [
        "batter_runs",
        "extra_runs",
        "total_runs",
    ]
].describe()
```

```python
invalid_run_rows = deliveries_df[
    deliveries_df["total_runs"]
    != (
        deliveries_df["batter_runs"]
        + deliveries_df["extra_runs"]
    )
]

print(
    "Invalid run rows:",
    len(invalid_run_rows)
)
```

#### Basic Dataset Summary

```python
summary = {
    "total_matches": matches_df["match_id"].nunique(),
    "total_deliveries": len(deliveries_df),
    "total_player_records": len(players_df),
    "unique_players": players_df["player_name"].nunique(),
    "unique_teams": players_df["team"].nunique(),
    "first_match_date": matches_df["match_date"].min(),
    "latest_match_date": matches_df["match_date"].max(),
}

summary
```
