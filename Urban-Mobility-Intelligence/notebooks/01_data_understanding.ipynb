#### <b>STEP 20: Shared Missing-Value Pattern</b>

Several columns contain exactly the same number of missing values.
Before deciding how to handle these missing values, we need to
determine whether the missing values occur in the same records.

This will help us understand whether the missingness is associated
with a particular subset of trips.

```python
shared_missing_columns = [
    "passenger_count",
    "congestion_surcharge",
    "store_and_fwd_flag",
    "RatecodeID",
    "Airport_fee"
]

shared_missing_mask = df[shared_missing_columns].isnull().all(axis=1)

print(
    "Records missing all five fields:",
    f"{shared_missing_mask.sum():,}"
)

print(
    "Percentage:",
    f"{shared_missing_mask.mean() * 100:.2f}%"
)
```

```python
partial_missing_mask = (
    df[shared_missing_columns].isnull().sum(axis=1).between(1, 4)
)

print(
    "Records with partial missingness:",
    f"{partial_missing_mask.sum():,}"
)
```

#### <b>STEP 23: BOUNDARY DATES INVESTIGATION</b>

```python
january_start = pd.Timestamp("2025-01-01")
february_start = pd.Timestamp("2025-02-01")

outside_january = (
    (df["tpep_pickup_datetime"] < january_start) |
    (df["tpep_pickup_datetime"] >= february_start)
)

print(
    "Pickup records outside January 2025:",
    f"{outside_january.sum():,}"
)

print(
    "Percentage:",
    f"{outside_january.mean() * 100:.4f}%"
)
```

```python
df.loc[
    outside_january,
    [
        "tpep_pickup_datetime",
        "tpep_dropoff_datetime",
        "VendorID",
        "PULocationID",
        "DOLocationID",
        "fare_amount",
        "total_amount"
    ]
].head(10)
```

```python
outside_january_dropoff = (
    (df["tpep_dropoff_datetime"] < january_start) |
    (df["tpep_dropoff_datetime"] >= february_start)
)

print(
    "Drop-off records outside January 2025:",
    f"{outside_january_dropoff.sum():,}"
)

print(
    "Percentage:",
    f"{outside_january_dropoff.mean() * 100:.4f}%"
)
```

#### <b>Phase 1 — Data Understanding Findings</b>

##### <b>1. Dataset Overview</b>

The January 2025 NYC Yellow Taxi dataset contains 3,475,226
trip records and 20 columns.

The source file is stored in Parquet format and has a size of
approximately 0.06 GB. After loading into Pandas, the DataFrame
uses approximately 493.47 MB of memory.

The dataset contains trip timestamps, passenger information,
pickup and drop-off locations, trip distance, payment information,
fare components, and total trip charges.

---

##### <b>2. Data Completeness</b>

Five columns contain missing values:

- passenger_count
- RatecodeID
- store_and_fwd_flag
- congestion_surcharge
- Airport_fee

Each contains 540,149 missing values, representing approximately
15.54% of the dataset.

All five fields are missing together for the same records, and
there are no records with partial missingness across these fields.

The records with this shared missingness have payment_type = 0.

Missing records will not be removed at this stage. Their business
meaning will be investigated during the data-cleaning phase.

---

##### <b>3. Duplicate Records</b>

The dataset contains:

- 0 exact duplicate records
- 0.00% duplicate rate

Therefore, no duplicate-removal action is currently required.

---

##### <b>4. Temporal Coverage</b>

The pickup timestamps range from:

2024-12-31 20:47:55

to:

2025-02-01 00:00:44

Only 22 pickup records fall outside the January 2025 calendar
period.

Drop-off timestamps range from:

2024-12-18 07:52:40

to:

2025-02-01 23:44:11

There are 1,811 drop-off records outside the January 2025 period.

These records will be investigated before any filtering is applied,
because a January pickup may legitimately have a February drop-off.

---

##### <b>5. Location Coverage</b>

The dataset contains:

- 261 unique pickup locations
- 260 unique drop-off locations

Pickup and drop-off locations are represented using TLC Taxi Zone
identifiers.

These identifiers will later be connected to the official taxi
zone lookup data.

---

##### <b>6. Numerical Data Quality</b>

Several numerical variables contain potentially anomalous values.

`Passenger Count`

- 24,656 records have zero passengers.
- No negative passenger counts were observed.
- 540,149 records have missing passenger counts.

`Trip Distance`

- 90,893 records have zero distance.
- No negative distances were observed.
- Maximum distance is 276,423.57 miles.

The extremely high maximum indicates the presence of severe
outliers that require investigation.

`Fare Amount`

- 1,398 records have zero fare.
- 144,118 records have negative fares.
- Maximum fare is 863,372.12.

The extreme maximum and negative values require investigation
before model development.

`Tip Amount`

- 1,118,008 records have zero tips.
- 124 records have negative tips.
- Maximum tip is 400.

Zero tips are potentially valid, while negative tips require
investigation.

`Tolls Amount`

- 3,259,590 records have zero tolls.
- 4,559 records have negative toll values.
- Maximum toll amount is 170.94.

Zero tolls can be legitimate because not every trip uses toll roads.

`Total Amount`

- 559 records have zero total amount.
- 63,037 records have negative total amounts.
- Maximum total amount is 863,380.37.

The extreme values require investigation before the dataset is
used for modeling.

---

##### <b>7. Trip Duration</b>

Trip duration was derived temporarily from pickup and drop-off
timestamps for data-understanding purposes.

1,377 trips have durations greater than three hours.

These trips will be investigated during data cleaning rather than
automatically removed.

---

##### <b>8. Initial Business Observations</b>

The raw NYC taxi data contains several characteristics that can
materially affect business analysis and machine-learning models.

The most important observations are:

1. Taxi trips are highly concentrated around short distances,
   as indicated by the difference between mean and median trip
   distance.

2. Extreme trip-distance and fare values can heavily distort
   averages.

3. Missing values are structured and associated with
   payment_type = 0 rather than appearing randomly across
   individual columns.

4. The dataset contains a small number of trips crossing the
   calendar boundaries of the January dataset.

5. Taxi demand analysis will require aggregation by time and
   location.

6. Fare prediction will require careful treatment of anomalous
   financial and trip records.

7. Location IDs will need to be mapped to taxi-zone information
   before meaningful geographic analysis.

---
